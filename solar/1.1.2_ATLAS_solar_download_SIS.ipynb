```python
import eumdac
import fnmatch
import shutil
import datetime
import time
import os

# Authenticate
credentials = (c_key, c_secret)

# We're passing our consumer_key and consumer_secret to eumdac's AccessToken function
token = eumdac.AccessToken(credentials)

print(f"This token '{token}' expires {token.expiration}")

datastore = eumdac.DataStore(token)
datatailor = eumdac.DataTailor(token)

# Selecting the collection and product.
selected_collection = datastore.get_collection('EO:EUM:DAT:0863')

# Set sensing start and end time
start = datetime.datetime(2023, 10, 1, 9, 0)
end = datetime.datetime(2023, 10, 3, 9, 0)
```

List of the available data sets: we are interested in the "SISdm"

```python
selected_products = selected_collection.search(dtstart = start, dtend = end)
print(f'Found Datasets: {selected_products.total_results} datasets for the given time range')
for product in selected_products:
	print(str(product))
```

```python
output_dir = '../data/dwd_downloads/'
os.makedirs(output_dir, exist_ok=True)

for product in selected_products:
    product_id = str(product)
    if product_id.startswith("SISdm"):
        print(f"⬇️  Downloading: {product_id}")
        with product.open() as fsrc, open(os.path.join(output_dir, product_id + ".nc"), 'wb') as fdst:
            shutil.copyfileobj(fsrc, fdst)
        print(f"✅ Saved: {product_id}.nc")
```
