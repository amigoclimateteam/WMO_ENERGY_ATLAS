```python
# ERA5-Land Data Download Script
# This notebook will help you download monthly data for surface solar radiation downwards,
# and 10m_u, 10m_v components of wind from the ERA5-Land dataset using the CDS API.

# Import necessary libraries
import cdsapi
import os
from datetime import datetime, timedelta

key = '3bf27ef9-17b4-4083-9039-ab99131fb5be' ## FILL WITH PERSONAL KEY
    
# Initialize the CDS API client
c = cdsapi.Client(url = 'https://cds.climate.copernicus.eu/api',
                        key = str(key), progress=False)
# Define the download function
def download_era5_land_monthly(variable, start_date, end_date, output_path, area=None):
    """
    Download ERA5-Land data one month at a time for a specified variable.
    Saves each month's data as a separate .nc file in the output path.
    
    Parameters:
    - variable: str, the ERA5 variable to download (e.g., 'surface_solar_radiation_downwards')
    - start_date: datetime, start date for the download
    - end_date: datetime, end date for the download
    - output_path: str, directory to save downloaded files
    - area: list, optional geographic bounding box [N, W, S, E]
    """
    current_date = start_date

    while current_date <= end_date:
        # Define the month and year for the current request
        year = current_date.year
        month = current_date.strftime("%m")
        
        # Define filename for the current month
        filename = f"{output_path}/{variable}_{year}_{month}.nc"
        os.makedirs(output_path, exist_ok=True)
        
        # Request parameters
        request_params = {
            'product_type': 'reanalysis',
            'format': 'netcdf',
            'download_format': 'unarchived',
            'variable': variable,
            'year': str(year),
            'month': month,
            'day': [f"{day:02d}" for day in range(1, 32)],
            'time': [f"{hour:02d}:00" for hour in range(24)]
        }
        
        if area:
            # Set the area for download [N, W, S, E] in degrees
            request_params['area'] = area

        # Make API request and download data
        print(f"Downloading {variable} data for {year}-{month}...")
        
        # NB. To download 100m wind data, you need to use reanalysis-era5 data source
        
        c.retrieve("reanalysis-era5-land", request_params, filename)
        
        # hourly ERA5 data → set dataset to 'reanalysis-era5-single-levels' 
        
        # monthly-averaged ERA5-Land data → set dataset to 'reanalysis-era5-land-monthly-means'
            #'product_type': 'monthly_averaged_reanalysis'
            #remove 'day' and 'time' fields, and set 'month': [MM], 'year': [YYYY];
        # monthly-averaged ERA5 data → use 'reanalysis-era5-monthly-means' with similar changes as above.
        
        print(f"Downloaded {variable} data to {filename}")
        
        # Move to the next month
        current_date += timedelta(days=31)
        current_date = current_date.replace(day=1)
```

```python
# Set download parameters
output_path = '../data/cds_downloads/' # Directory to save downloaded files

area = [23, -92, 7, -77]# Central America bounding box [North, West, South, East]

start_date = datetime(1992, 1, 1)  # Starting date
end_date = datetime(2020, 12, 31)  # Ending date

# Download 10m u-component of wind data month by month
download_era5_land_monthly(
    variable="10m_u_component_of_wind", #10m_v_component_of_wind
    start_date=start_date,
    end_date=end_date,
    output_path=output_path,
    area=area
)
```
