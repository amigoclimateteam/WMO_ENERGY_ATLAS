```python
### REQUIRES conda_solar_wind enviroment from conda_solar_wind.yml

from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
import numpy as np
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
import os
```

For the wind, we downscale the two components of the wind and wind speed separately with three dedicated models.

Then, we use the downscaled output of the two components to compute the downscaled wind direction.

```python
# Methods that are used below.

def round_coords(xdf, digits = None):
    """
    Rounds the coordinates of the xarray DataArray or Dataset to the specified number of decimal places.
    Needed to prevent issues due to excess decimal digits in coordinates, leading to grid mismatches.

    Parameters:
    xdf (xarray.DataArray or xarray.Dataset): The data containing 'latitude' and 'longitude' to be rounded.
    digits (int, optional): The number of decimal places to round to. Defaults to 1 if not provided.

    Returns:
    xarray.DataArray or xarray.Dataset: The data with rounded coordinates.
    """
    if not digits:
        # Round latitude and longitude to 1 decimal place by default
        xdf['latitude'] = xdf.latitude.round(1).astype('float32')
        xdf['longitude'] = xdf.longitude.round(1).astype('float32')
    else:
        # Round latitude and longitude to the specified number of decimal places
        xdf['latitude'] = xdf.latitude.round(digits).astype('float32')
        xdf['longitude'] = xdf.longitude.round(digits).astype('float32')    
    return xdf

def make_dataset_era5land(target_var, orography, aspect, monthly_avg = False):
    """
    Creates a dataset by merging a target variable, orography, and aspect data, optionally computing monthly averages.

    Parameters:
    target_var (xarray.DataArray or xarray.Dataset): The target variable data.
    orography (xarray.DataArray or xarray.Dataset): Orography data.
    aspect (xarray.DataArray or xarray.Dataset): Aspect data.
    monthly_avg (bool, optional): Whether to compute monthly averages. Defaults to False.

    Returns:
    pandas.DataFrame: The merged dataset in a DataFrame format, with rounded coordinates.
    """
    for xdf in [target_var, orography, aspect]:
        xdf = round_coords(xdf)
        
    # Merge the datasets into one
    xdf_merged = xr.merge([orography, aspect, target_var])
    
    if monthly_avg:
        # Compute the monthly average if specified
        xdf_merged = xdf_merged.groupby(xdf_merged.time.dt.month).mean()
        
    # Convert to DataFrame and drop NaN values
    merged = xdf_merged.to_dataframe().dropna()
    merged = merged.reset_index()
    
    # Round coordinates for final output
    merged['longitude'] = merged['longitude'].round(1)
    merged['latitude'] = merged['latitude'].round(1)
    return merged

def make_dataset_glo90(orography, aspect):
    """
    Creates a dataset by merging GLO-90 orography and aspect data, with coordinates rounded to 6 decimal places (as 90 m correspond to about 0.000808°).

    Parameters:
    orography (xarray.DataArray or xarray.Dataset): Orography data.
    aspect (xarray.DataArray or xarray.Dataset): Aspect data.

    Returns:
    pandas.DataFrame: The merged dataset in a DataFrame format, with NaN values for zeroes and reset index.
    """
    for xdf in [orography, aspect]:
        xdf = round_coords(xdf, digits = 6)
        
    # Merge the GLO-90 datasets
    merged_glo90 = xr.merge([orography, aspect])
    
    # Convert to DataFrame and drop NaN values
    merged_glo90 = merged_glo90.to_dataframe().dropna()
    merged_glo90 = merged_glo90.reset_index()

    return merged_glo90

def era5scaler(xdf, features, target):
    """
    Scales the features and target variables of the ERA5Land dataset using StandardScaler.

    Parameters:
    xdf (pandas.DataFrame): The dataset containing the features and target.
    features (list of str): List of feature column names to scale.
    target (str): The target column name to scale.

    Returns:
    tuple: The scaler for the target, scaled feature array, and scaled target array.
    """
    featurescaler = StandardScaler()
    # Scale the features
    X_scaled = featurescaler.fit_transform(xdf[features])

    targetscaler = StandardScaler()
    # Scale the target
    y_scaled = targetscaler.fit_transform(xdf[[target]]).ravel()
    
    return targetscaler, X_scaled, y_scaled

def glo90scaler(xdf, features):
    """
    Scales the features of the GLO-90 dataset using StandardScaler.

    Parameters:
    xdf (pandas.DataFrame): The dataset containing the features.
    features (list of str): List of feature column names to scale.

    Returns:
    numpy.ndarray: The scaled feature array.
    """
    scaler_glo90 = StandardScaler()
    # Scale the features
    X_scaled = scaler_glo90.fit_transform(xdf[features])
    return X_scaled
```

```python
def WindCalc(xdf_wind):
    """
    Calculates wind speed and direction from downscaled u10 and v10 wind components.

    Parameters:
    xdf_wind (pandas.DataFrame or xarray.Dataset): The dataset containing 'u10_downscaled' and 'v10_downscaled'.

    Returns:
    pandas.DataFrame or xarray.Dataset: The dataset with additional columns 'wds' (wind speed) and 'direction_deg' (wind direction in degrees).
    """
    # Calculate wind speed (wds) using the Pythagorean theorem
    xdf_wind['wds_downscaled'] = ((xdf_wind[f'u10_downscaled']**2 + xdf_wind[f'v10_downscaled']**2)**(1/2))
    # Calculate wind direction in degrees, converting from radians
    #Default: from the North, counterclockwise turn
    xdf_wind['dir_downscaled'] = (np.rad2deg(np.arctan2(xdf_wind[f'v10_downscaled'], xdf_wind[f'u10_downscaled'])) + 360) % 360
    
    #Cuba has the stations measurements in clockwise turn from the North - conversion needed!
    xdf_wind['dir_downscaled'] = (270 - xdf_wind['dir_downscaled']) % 360
    return xdf_wind
```

### Preparation of the input dataset with ERA5Land features

```python
country = 'cuba' ## FILL WITH THE COUNTRY NAME e.g. croatia, cuba
```

```python
data_path = f'../data/processed_data/{country}/' ## FILL WITH THE PATH of ERA5LAND data
output_path = f'../data/downscaled_data/{country}/' ## FILL WITH THE PATH of desidered OUTPUT PATH

output_dir = os.path.dirname(output_path)
os.makedirs(output_dir, exist_ok=True)
```

```python
def era5land_preparation(data_path,country, component):
    ### Load ERA5Land features
    comp = xr.open_mfdataset(f'{data_path}{component}10_1991-01_2020-12_processed.nc')
    # ERA5Land-derived DEM and aspect datasets
    orography = xr.open_dataset(f'../DEMdata/{country}/era5land_orography_{country}.nc')
    aspect = xr.open_dataset(f'../DEMdata/{country}/era5land_aspect_{country}.nc')
    ### Make the datasets with longitude, latitude, month of the year, altitude, aspect, and surface downward solar radiation.
    ### The datasets consider the average of the target wind variable for each month of the year across the input dataset.

    comp_era5l_merged = make_dataset_era5land(comp, orography, aspect, monthly_avg = True)

    return comp_era5l_merged
```

```python
u10_era5l_merged = era5land_preparation(data_path, country, 'u')
v10_era5l_merged = era5land_preparation(data_path, country, 'v')
```

```python
month = 1
```

```python
# Training the downscaling model on ERA5Land features
def training_ds(comp_era5l_merged, month, component):
    ### Select a target month (here is January); and then drop the month feature.
    comp_era5l_merged = comp_era5l_merged.loc[comp_era5l_merged.month == month,:].drop('month', axis = 1)
    ### Scaling features and target
    features = comp_era5l_merged.columns.drop([f'{component}10','spatial_ref'])#,'number'])
    target = f'{component}10'

    ## This method returns the scaler (needed to turn the predictions of the model on GLO-90 features into downscaled values),
    # and the scaled input feature set and target
    targetscaler_era5, era5l_X, era5l_y = era5scaler(comp_era5l_merged, features, target)
    
    # Fit the downscaling model. The type of model is a neural network regressor (multilayer perceptron)
    regr = MLPRegressor(random_state=1, max_iter = 1000).fit(era5l_X, era5l_y)
    
    return features, targetscaler_era5, era5l_X, era5l_y, regr
```

```python
features_u10, targetscaler_era5_u10, era5l_X_u10, era5l_y_u10, regr_u10 = training_ds(u10_era5l_merged, month,'u')
features_v10, targetscaler_era5_v10, era5l_X_v10, era5l_y_v10, regr_v10 = training_ds(v10_era5l_merged, month,'v')
```

### Prepare the dataset with GLO-90 features

```python
def opening_glo90(country):
    ### Prepare the GLO-90 features
    orography_glo90 = xr.open_dataset(f'../DEMdata/{country}/glo90_orography_{country}.nc')
    aspect_glo90 = xr.open_dataset(f'../DEMdata/{country}/glo90_aspect_{country}.nc')
    dataset_glo90 = make_dataset_glo90(orography_glo90,aspect_glo90)
    dataset_glo90 = dataset_glo90.rename(columns = {'band_data':'aspect'})
    return dataset_glo90
```

```python
dataset_glo90 = opening_glo90(country)
```

### Downscaling: use the model on GLO-90 features

```python
# Downscaling: use the model on GLO-90 features
def apply_downscaling(dataset_glo90, regr, targetscaler_era5, features,month,country,component):
    ### Scale GLO-90 features
    glo90_X = glo90scaler(dataset_glo90, features)
    
    ### Calculate the downscaled values
    downscaled_mlr_std = regr.predict(glo90_X) # These values are standardised, make them go back to actual values
    downscaled_mlr = targetscaler_era5.inverse_transform(downscaled_mlr_std.reshape(-1,1)) # Using the targetscaler
    dataset_glo90[f'{component}10_downscaled'] = downscaled_mlr 
    output_ds = dataset_glo90.set_index(['latitude', 'longitude'])[[f'{component}10_downscaled']].to_xarray()
    
    return output_ds
```

```python
output_ds_u10 = apply_downscaling(dataset_glo90, regr_u10, targetscaler_era5_u10, features_u10,month,country,'u')
output_ds_v10 = apply_downscaling(dataset_glo90, regr_v10, targetscaler_era5_v10, features_v10,month,country,'v')
```

```python
# Merge the downscaled u and v component
wind_output = xr.merge([output_ds_u10, output_ds_v10])

# Calculate wind direction using the method
wind_output = WindCalc(wind_output)
```

```python
wind_output.to_netcdf(f'{output_path}10m_wind_downscaled_{country}_m{month}.nc')
```
