```python
# GLOFAS River Discharge Data Download Script
# This notebook will help you download monthly river discharge data from GLOFAS
# using the CDS API.

# Import necessary libraries
import cdsapi
import os
from datetime import datetime, timedelta


key = '3bf27ef9-17b4-4083-9039-ab99131fb5be' ## FILL WITH PERSONAL KEY
    
# Initialize the CDS API client
c = cdsapi.Client(url = 'https://ewds.climate.copernicus.eu/api',
                        key = str(key), progress=False)

# Define the download function
def download_glofas_river_discharge_monthly(start_date, end_date, output_path, area=None):
    """
    Download GLOFAS river discharge data one month at a time.
    Saves each month's data as a separate .nc file in the output path.
    
    Parameters:
    - start_date: datetime, start date for the download
    - end_date: datetime, end date for the download
    - output_path: str, directory to save downloaded files
    - area: list, optional geographic bounding box [N, W, S, E]
    """
    current_date = start_date

    while current_date <= end_date:
        # Define the month and year for the current request
        year = current_date.year
        month = current_date.strftime("%m")
        
        # Define filename for the current month
        filename = f"{output_path}/river_discharge_{year}_{month}.grib"
        os.makedirs(output_path, exist_ok=True)
        
        request_params = {
            "system_version": ["version_4_0"],
            "hydrological_model": ["lisflood"],
            "product_type": ["consolidated"],
            "variable": ["river_discharge_in_the_last_24_hours"],
            "hyear": str(year),
            "hmonth": month,
            "hday": [f"{day:02d}" for day in range(1, 32)],
            "data_format": "grib2",
            "download_format": "unarchived"
        }
        
        if area:
            # Set the area for download [N, W, S, E] in degrees
            request_params['area'] = area

        # Make API request and download data
        print(f"Downloading river discharge data for {year}-{month}...")
        c.retrieve("cems-glofas-historical", request_params, filename)
        print(f"Downloaded river discharge data to {filename}")
        
        # Move to the next month
        current_date += timedelta(days=31)
        current_date = current_date.replace(day=1)
```

```python

# Set download parameters
output_path = f'../data/glofas_downloads'  # Directory to save downloaded files
area = [-5, 10, -35, 55]  # SADC region bounding box [North, West, South, East]
start_date = datetime(1992, 1, 1)  # Starting date
end_date = datetime(2020, 12, 31)  # Ending date

# Download river discharge data month by month
download_glofas_river_discharge_monthly(
    start_date=start_date,
    end_date=end_date,
    output_path=output_path,
    area=area
)
```
